# **A/B Testing Analysis**

```python
# Sample data for demonstration
data = {
    'category': ['Sneakers', 'Sneakers', 'Boots', 'Boots', 'Sneakers', 'Boots', 'Sandals', 'Sneakers', 'Sandals', 'Boots'],
    'sales': [100, 150, 200, 250, 180, 220, 130, 160, 140, 260],
    'reviews_count': [10, 15, 25, 30, 20, 22, 8, 12, 10, 35],
    'average_rating': [4.5, 4.6, 4.3, 4.8, 4.5, 4.4, 4.0, 4.7, 4.5, 4.2]
}

# Create a DataFrame
df_sales = pd.DataFrame(data)

# Analyze sales data by category
results = df_sales.groupby('category').agg({
    'sales': 'sum',
    'reviews_count': 'sum',
    'average_rating': 'mean'
}).reset_index()

# Display the results
print("A/B Testing Results by Shoe Category:")
print(results)
```

```python
import pandas as pd

# Load the data
df = pd.read_csv('adidas_usa.csv')

# Check the columns
print("Available columns in the DataFrame:")
print(df.columns)

# Assume we want to segment by product category and color, as well as sales data
# Create a new DataFrame focusing on relevant attributes
segmentation_data = df[['category', 'color', 'sku', 'selling_price', 'availability', 'average_rating', 'reviews_count']]

# Group by category and color, and calculate relevant metrics
segmented_data = (
    segmentation_data
    .groupby(['category', 'color'])
    .agg(
        total_sales=('sku', 'nunique'),  # Count unique SKUs sold
        average_price=('selling_price', 'mean'),  # Average selling price
        average_rating=('average_rating', 'mean'),  # Average rating
        total_reviews=('reviews_count', 'sum')  # Total reviews
    )
    .reset_index()
)

# Display the segmented data
print(segmented_data)

# Optional: Further segment based on average price or rating
# For instance, you could segment by price brackets
price_bins = [0, 50, 100, 150, 200, float('inf')]
price_labels = ['Under $50', '$50-$100', '$100-$150', '$150-$200', 'Above $200']

segmented_data['price_segment'] = pd.cut(segmented_data['average_price'], bins=price_bins, labels=price_labels)

# Display the segmented data with price segments
print(segmented_data)

# Analyzing segment characteristics
for category in segmented_data['category'].unique():
    print(f"\nSegment Analysis for Category: {category}")
    category_data = segmented_data[segmented_data['category'] == category]
    print(category_data[['color', 'total_sales', 'average_price', 'average_rating', 'total_reviews']])
```

```python
import pandas as pd

# Sample Data: Create a DataFrame for demonstration
data = {
    'year': [2022, 2023],
    'sales': [10000, 12000]
}

# Load the dataset into a DataFrame
df = pd.DataFrame(data)

# Display the original DataFrame
print("Sales Data:")
print(df)

# Sort the DataFrame by 'year' to ensure correct order for growth calculation
df = df.sort_values(by='year')

# Calculate the YOY Growth
# Calculate percentage change and multiply by 100 to convert it to percentage
df['YOY Growth (%)'] = df['sales'].pct_change() * 100

# Display the DataFrame with YOY Growth
print("\nSales Data with YOY Growth:")
print(df)
```
